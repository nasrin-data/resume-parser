```python
def calculate_weighted_score(
    candidate_skills,
    required_skills,
    preferred_skills,
    experience_score,
    education_score
):
    candidate_skills = [skill.lower() for skill in candidate_skills]
if required_skills:
        matched_required = sum(
            1 for skill in required_skills
            if skill.lower() in candidate_skills
        )
        required_score = (matched_required / len(required_skills)) * 45
    else:
        required_score = 0
    if preferred_skills:
        matched_preferred = sum(
            1 for skill in preferred_skills
            if skill.lower() in candidate_skills
        )
        preferred_score = (matched_preferred / len(preferred_skills)) * 10
    else:
        preferred_score = 0
    experience_points = (experience_score / 100) * 20
    education_points = (education_score / 100) * 10

    similarity_points = 0

    total_score = (
        required_score
        + preferred_score
        + experience_points
        + education_points
        + similarity_points
    )

    return round(total_score, 2)
```

```python
import resume_parser
import re
from datetime import datetime
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

required_skills = ["Python", "SQL", "Excel", "Data Analysis"]

preferred_skills = [
    "Power BI", "Tableau", "Machine Learning"
]

job_description = """
Junior Data Analyst.
Required skills: Python, SQL, Excel, data cleaning and data analysis.
Preferred skills: Power BI, Tableau, machine learning and data visualization.
Experience: 0 to 2 years, including internships and projects.
Education: Bachelor's degree in Statistics, Computer Science,
Economics, Business or a related field.
"""

resumes = [
    "/content/sample_resumes/Candidate_Aarav_Mehta.pdf",
    "/content/sample_resumes/Candidate_Priya_Sharma.pdf",
    "/content/sample_resumes/Candidate_Rohan_Iyer.pdf",
    "/content/sample_resumes/Candidate_Fatima_Khan.pdf",
    "/content/sample_resumes/Candidate_Kabir_Singh.pdf"
]

def get_skill_points(candidate_skills, job_skills, maximum_points):
    candidate_skills = [s.lower().strip() for s in candidate_skills]

    if not job_skills:
        return 0

    matched = sum(
        1 for skill in job_skills
        if skill.lower().strip() in candidate_skills
    )

    return (matched / len(job_skills)) * maximum_points

def get_experience_score(text):
    months = {
        "jan": 1, "january": 1,
        "feb": 2, "february": 2,
        "mar": 3, "march": 3,
        "apr": 4, "april": 4,
        "may": 5,
        "jun": 6, "june": 6,
        "jul": 7, "july": 7,
        "aug": 8, "august": 8,
        "sep": 9, "sept": 9, "september": 9,
        "oct": 10, "october": 10,
        "nov": 11, "november": 11,
        "dec": 12, "december": 12
    }

    date_pattern = (
        r"\b(Jan(?:uary)?|Feb(?:ruary)?|Mar(?:ch)?|Apr(?:il)?|May|"
        r"Jun(?:e)?|Jul(?:y)?|Aug(?:ust)?|Sep(?:t(?:ember)?)?|"
        r"Oct(?:ober)?|Nov(?:ember)?|Dec(?:ember)?)"
        r"\s+(\d{4})\s*[-–to]+\s*"
        r"(Jan(?:uary)?|Feb(?:ruary)?|Mar(?:ch)?|Apr(?:il)?|May|"
        r"Jun(?:e)?|Jul(?:y)?|Aug(?:ust)?|Sep(?:t(?:ember)?)?|"
        r"Oct(?:ober)?|Nov(?:ember)?|Dec(?:ember)?)"
        r"\s+(\d{4})"
    )

    total_months = 0

    for match in re.finditer(date_pattern, text, re.IGNORECASE):
        start_month = months[match.group(1).lower()[:3]]
        start_year = int(match.group(2))
        end_month = months[match.group(3).lower()[:3]]
        end_year = int(match.group(4))

        duration = (
            (end_year - start_year) * 12
            + end_month - start_month + 1
        )

        if duration > 0:
            total_months += duration


    total_months = min(total_months, 24)

    return (total_months / 24) * 100


def get_education_score(text):
    text = text.lower()

    degree_words = [
        "b.sc", "b.s.", "b.tech", "b.e.",
        "bachelor", "b.com", "bca", "bba",
        "m.sc", "m.tech", "master", "mca"
    ]

    if any(degree in text for degree in degree_words):
        return 100

    return 0


candidates = []

for resume in resumes:
    text = resume_parser.read_resume(resume)
    details = resume_parser.extract_details(text)

    name = details.get("Name", "Not found")
    email = details.get("Email", "Not found")
    skills = details.get("Skills", [])

    required_points = get_skill_points(
        skills, required_skills, 45
    )

    preferred_points = get_skill_points(
        skills, preferred_skills, 10
    )

    experience_percentage = get_experience_score(text)
    experience_points = experience_percentage * 0.20

    education_percentage = get_education_score(text)
    education_points = education_percentage * 0.10


    vectorizer = TfidfVectorizer(stop_words="english")

    vectors = vectorizer.fit_transform([
        text, job_description
    ])

    similarity_percentage = (
        cosine_similarity(vectors[0:1], vectors[1:2])[0][0]
        * 100
    )

    similarity_points = similarity_percentage * 0.15

    total_score = (
        required_points
        + preferred_points
        + experience_points
        + education_points
        + similarity_points
    )

    candidates.append({
        "Name": name,
        "Email": email,
        "Skills": skills,
        "Required": round(required_points, 2),
        "Preferred": round(preferred_points, 2),
        "Experience": round(experience_points, 2),
        "Education": round(education_points, 2),
        "Similarity": round(similarity_points, 2),
        "Score": round(total_score, 2)
    })


candidates.sort(
    key=lambda candidate: candidate["Score"],
    reverse=True
)

print("RESUME SCREENING RESULTS")
print("=" * 75)

for rank, candidate in enumerate(candidates, start=1):
    print(f"\nRank: {rank} | {candidate['Name']}")
    print("Email:", candidate["Email"])
    print("Skills:", ", ".join(candidate["Skills"]))
    print("Required skills:", candidate["Required"], "/ 45")
    print("Preferred skills:", candidate["Preferred"], "/ 10")
    print("Experience:", candidate["Experience"], "/ 20")
    print("Education:", candidate["Education"], "/ 10")
    print("Job similarity:", candidate["Similarity"], "/ 15")
    print("TOTAL SCORE:", candidate["Score"], "/ 100")
    print("-" * 75)
```

```python
%%writefile app.py

import streamlit as st
import resume_parser
import tempfile
import os

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

st.set_page_config(page_title="CloudRiq Resume Screening")

st.title("CloudRiq Resume Screening")
st.write("Upload resumes and compare them against a job description.")

job_description = st.text_area(
    "Enter the job description",
    height=180
)

uploaded_files = st.file_uploader(
    "Upload candidate resumes (PDF)",
    type=["pdf"],
    accept_multiple_files=True
)

required_text = st.text_input(
    "Required skills (comma-separated)",
    "Python, SQL, Excel, Data Analysis"
)

preferred_text = st.text_input(
    "Preferred skills (comma-separated)",
    "Power BI, Tableau, Machine Learning"
)

if st.button("Screen Resumes"):
    if not job_description.strip():
        st.warning("Please enter a job description.")
    elif not uploaded_files:
        st.warning("Please upload at least one PDF resume.")
    else:
        required_skills = [
            s.strip() for s in required_text.split(",") if s.strip()
        ]
        preferred_skills = [
            s.strip() for s in preferred_text.split(",") if s.strip()
        ]

        results = []

        for uploaded_file in uploaded_files:
            temp_path = None

            try:
                with tempfile.NamedTemporaryFile(
                    suffix=".pdf", delete=False
                ) as temp_file:
                    temp_file.write(uploaded_file.getvalue())
                    temp_path = temp_file.name

                resume_text = resume_parser.read_resume(temp_path)
                details = resume_parser.extract_details(resume_text)

                skills = details.get("Skills", [])
                if isinstance(skills, str):
                    skills = [s.strip() for s in skills.split(",")]
                matched_required = sum(
                    1 for skill in required_skills
                    if skill.lower() in [
                        s.lower() for s in skills
                    ]
                )
                required_points = (
                    matched_required / len(required_skills) * 45
                    if required_skills else 0
                )
                matched_preferred = sum(
                    1 for skill in preferred_skills
                    if skill.lower() in [
                        s.lower() for s in skills
                    ]
                )
                preferred_points = (
                    matched_preferred / len(preferred_skills) * 10
                    if preferred_skills else 0
                )

                if resume_text.strip():
                    vectorizer = TfidfVectorizer(
                        stop_words="english"
                    )
                    vectors = vectorizer.fit_transform([
                        resume_text, job_description
                    ])
                    similarity = cosine_similarity(
                        vectors[0:1], vectors[1:2]
                    )[0][0] * 15
                else:
                    similarity = 0
                experience_points = 0
                education_points = 0

                total = (
                    required_points
                    + preferred_points
                    + experience_points
                    + education_points
                    + similarity
                )

                results.append({
                    "Candidate": details.get("Name", "Not found"),
                    "Email": details.get("Email", "Not found"),
                    "Skills": ", ".join(skills),
                    "Required skills": round(required_points, 2),
                    "Preferred skills": round(preferred_points, 2),
                    "Experience": experience_points,
                    "Education": education_points,
                    "Similarity": round(similarity, 2),
                    "Score": round(total, 2)
                })

            except Exception as error:
                st.error(
                    f"Could not process {uploaded_file.name}: {error}"
                )

            finally:
                if temp_path and os.path.exists(temp_path):
                    os.remove(temp_path)

        results.sort(
            key=lambda item: item["Score"],
            reverse=True
        )

        if results:
            st.subheader("Candidate Ranking")

            for rank, candidate in enumerate(results, start=1):
                st.markdown(f"### Rank {rank}: {candidate['Candidate']}")
                st.write("Email:", candidate["Email"])
                st.write("Skills:", candidate["Skills"])
                st.write("Required skills:", candidate["Required skills"], "/ 45")
                st.write("Preferred skills:", candidate["Preferred skills"], "/ 10")
                st.write("Experience:", candidate["Experience"], "/ 20")
                st.write("Education:", candidate["Education"], "/ 10")
                st.write("Job similarity:", candidate["Similarity"], "/ 15")
                st.write("*Total score:", candidate["Score"], "/ 100*")
                st.divider()

            st.caption(
                "Prototype only: experience and education scoring "
                "are not implemented yet. Scores must not be used "
                "for automatic rejection or hiring decisions."
            )
```

```python
%%writefile resume_parser_app.py
import streamlit as st

st.title("CloudRiq Resume Screening Web App")
st.subheader("Week 3 & 4: Text Tokenization & Scoring Engine")
text_input = st.text_area("Paste Resume Text Content Here:", height=200, placeholder="Paste text content...")

if st.button("Run Tokenizer & Calculate Score"):
    if text_input.strip():
        clean_text = text_input.lower()
        found_required = []
        if "python" in clean_text:
            found_required.append("Python")
        if "pandas" in clean_text:
            found_required.append("Pandas")
        if "power bi" in clean_text:
            found_required.append("Power BI")
        req_score = len(found_required) * 15.0
        found_preferred = []
        if "sql" in clean_text:
            found_preferred.append("SQL")
        if "excel" in clean_text:
            found_preferred.append("Excel")
        pref_score = len(found_preferred) * 5.0
        exp_score = 0.0
        detected_exp = "0 years"
        if "1 year" in clean_text or "1 yrs" in clean_text or "1 yr" in clean_text:
            exp_score = 10.0
            detected_exp = "1 year"
        elif "2 year" in clean_text or "2 yrs" in clean_text or "2 yr" in clean_text:
            exp_score = 20.0
            detected_exp = "2 years"
        elif "3 year" in clean_text or "3 yrs" in clean_text or "3 yr" in clean_text:
            exp_score = 20.0
            detected_exp = "3+ years"
        grand_total = req_score + pref_score + exp_score
        st.success("Successfully processed text tokens and calculated matching metrics!")

        st.markdown(f"### 📊 Overall Candidate Match Score: *{grand_total}% / 75%*")
        st.write(f"ℹ️ Breakdown: Required Skills: {req_score}% | Preferred Skills: {pref_score}% | Experience: {exp_score}%")
        st.info(f"💼 *Detected Candidate Experience:* {detected_exp}")

        col1, col2 = st.columns(2)
        with col1:
            st.write("✅ *Required Skills Found:*", found_required if found_required else "None")
        with col2:
            st.write("⭐ *Preferred Skills Found:*", found_preferred if found_preferred else "None")
        with st.expander("See Extracted Tokens List"):
            st.write(clean_text.split())
    else:
        st.warning("Please enter valid text to tokenize.")
```
